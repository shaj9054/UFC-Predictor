```python
df_clean = df[df['Winner'].isin(['Red', 'Blue'])]
df_clean = df_clean[df_clean['finish'].notnull()]
df_clean.shape
```

```python
finish_map = {
    'U-DEC': 'DEC',
    'S-DEC': 'DEC',
    'SUB': 'SUB',
    'KO/TKO': 'KO/TKO',
    'M-DEC': 'DEC',
    # ... you fill in the rest
}
df_clean['method'] = df_clean['finish'].map(finish_map)
df_clean['method'].value_counts(dropna=False)
```

```python
df_clean = df_clean[df_clean['method'].notnull()].copy()
df_clean.shape
```

```python
(df_clean['method'].value_counts(normalize=True)* 100).round(1)
```

```python
(df_clean['Winner'].value_counts(normalize=True)* 100).round(1)
```

```python
(df_clean.groupby('method')['Winner'].value_counts(normalize=True)* 100).round(1)
```
