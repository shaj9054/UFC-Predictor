```python
df_clean = df[df['Winner'].isin(['Red', 'Blue'])]
df_clean = df_clean[df_clean['finish'].notnull()]
df_clean.shape
```

```python
finish_map = {
    'U-DEC': 'DEC',
    'S-DEC': 'DEC',
    'SUB': 'SUB',
    'KO/TKO': 'KO/TKO',
    'M-DEC': 'DEC',
    # ... you fill in the rest
}
df_clean['method'] = df_clean['finish'].map(finish_map)
df_clean['method'].value_counts(dropna=False)
```

```python
r_cols = [c for c in df_clean.columns if c.startswith('R_')]
b_cols = [c for c in df_clean.columns if c.startswith('B_')]

print(len(r_cols), len(b_cols))
```

```python
r_stats = {c[2:] for c in r_cols}
b_stats = {c[2:] for c in b_cols}

paired = sorted(r_stats & b_stats)
print(len(paired))
print(paired)
```

```python
for s in paired:
    print(s)
```

```python
ranks = [s for s in paired if 'rank' in s.lower()]
print(len(ranks))
print(ranks)
```

```python
print(df_clean['R_match_weightclass_rank'].isnull().mean() * 100)
print(df_clean['R_Lightweight_rank'].isnull().mean() * 100)
```

```python
df_clean['R_match_weightclass_rank'].value_counts().sort_index()
```

```python
df_clean['R_is_ranked'] = df_clean['R_match_weightclass_rank'].notnull().astype(int)
df_clean['B_is_ranked'] = df_clean['B_match_weightclass_rank'].notnull().astype(int)

df_clean['R_rank_filled'] = df_clean['R_match_weightclass_rank'].fillna(20)
df_clean['B_rank_filled'] = df_clean['B_match_weightclass_rank'].fillna(20)
```

```python
df_clean['rank_dif'] = df_clean['B_rank_filled'] - df_clean['R_rank_filled']
```

```python
drop_these = ['fighter', 'odds', 'ev', 'Stance']
numeric_pairs = [s for s in paired if s not in ranks and s not in drop_these]

print(len(numeric_pairs))
print(numeric_pairs)
```

```python
for stat in numeric_pairs:
    df_clean[f'{stat}_dif'] = df_clean[f'R_{stat}'] - df_clean[f'B_{stat}']
```

```python
dif_cols = [c for c in df_clean.columns if c.endswith('_dif')]
print(len(dif_cols))
```
